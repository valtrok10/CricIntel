```python
import json
from pathlib import Path
import pandas as pd

raw_folder = Path("../data/raw")

json_files = list(raw_folder.glob("*.json"))

print("Number of matches:", len(json_files))
print(json_files[:5])
```

```python
records = []

for file_path in json_files:

    with open(file_path, "r", encoding="utf-8") as file:
        match = json.load(file)

    match_id = file_path.stem

    info = match["info"]

    dates = info.get("dates", [])
    match_date = dates[0] if dates else None

    venue = info.get("venue")
    city = info.get("city")

    teams = info.get("teams", [])

    for innings_number, innings in enumerate(match["innings"], start=1):

        batting_team = innings["team"]
        bowling_team = next(
    (team for team in teams if team != batting_team),
    None
)

        for over in innings["overs"]:

            over_number = over["over"]

            for delivery in over["deliveries"]:
                extras = delivery.get("extras", {})
                wides = extras.get("wides", 0)
                noballs = extras.get("noballs", 0)
                byes = extras.get("byes", 0)
                legbyes = extras.get("legbyes", 0)
                is_legal_delivery = wides == 0 and noballs == 0
                wickets = delivery.get("wickets", [])
                is_wicket = len(wickets) > 0;
                record = {
                    "match_id": match_id,
                    "date": match_date,
                    "city": city,
                    "venue": venue,
                    "innings": innings_number,
                    "batting_team": batting_team,
                    "bowling_team": bowling_team,
                    "over": over_number,
                    "wides": wides,
                    "noballs": noballs,
                    "byes": byes,
                    "legbyes": legbyes,
                    "is_legal_delivery": is_legal_delivery,
                    "is_wicket": is_wicket,
                    "delivery": delivery.get("actual_delivery"),
                    "batter": delivery["batter"],
                    "non_striker": delivery["non_striker"],
                    "bowler": delivery["bowler"],
                    "batter_runs": delivery["runs"]["batter"],
                    "extra_runs": delivery["runs"]["extras"],
                    "total_runs": delivery["runs"]["total"]
                    
                }

                records.append(record)
```

```python
df = pd.DataFrame(records)

df.head()
```

```python
# Convert date to datetime
df["date"] = pd.to_datetime(df["date"])

# Season
df["season"] = df["date"].dt.year

# Match phase
def get_phase(over):
    if over < 6:
        return "Powerplay"
    elif over < 15:
        return "Middle"
    else:
        return "Death"

df["phase"] = df["over"].apply(get_phase)

# Boundary delivery
df["is_boundary"] = df["batter_runs"].isin([4, 6])

# Dot ball
df["is_dot"] = df["total_runs"] == 0

# Ball faced by batter
df["ball_faced"] = (df["wides"] == 0).astype(int)
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
top_run_scorers = (
    df.groupby("batter")["batter_runs"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_run_scorers
```

```python
batter_stats = (
    df.groupby("batter")
      .agg(
          runs=("batter_runs", "sum"),
          balls=("ball_faced", "sum"),
          boundaries=("is_boundary", "sum"),
          dot_balls=("is_dot", "sum")
      )
      .reset_index()
)
```

```python
batter_stats["strike_rate"] = (
    batter_stats["runs"] / batter_stats["balls"] * 100
).round(2)

batter_stats["boundary_pct"] = (
    batter_stats["boundaries"] / batter_stats["balls"] * 100
).round(2)

batter_stats["dot_ball_pct"] = (
    batter_stats["dot_balls"] / batter_stats["balls"] * 100
).round(2)
```

```python
top10_analysis = (
    batter_stats
    .sort_values("runs", ascending=False)
    .head(10)
)

top10_analysis
```

```python
phase_stats = (
    df.groupby(["batter", "phase"])
      .agg(
          runs=("batter_runs", "sum"),
          balls=("ball_faced", "sum"),
          boundaries=("is_boundary", "sum"),
          dots=("is_dot", "sum")
      )
      .reset_index()
)

phase_stats["strike_rate"] = (
    phase_stats["runs"] / phase_stats["balls"] * 100
).round(2)

phase_stats["boundary_pct"] = (
    phase_stats["boundaries"] / phase_stats["balls"] * 100
).round(2)

phase_stats["dot_ball_pct"] = (
    phase_stats["dots"] / phase_stats["balls"] * 100
).round(2)
```

```python
import sqlite3

conn = sqlite3.connect("../sql/cricintel.db")

df.to_sql(
    "deliveries",
    conn,
    if_exists="replace",
    index=False
)

print("Database created!")
```

```python
query = """
SELECT COUNT(*) AS total_deliveries
FROM deliveries;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    batter,
    SUM(batter_runs) AS total_runs
FROM deliveries
GROUP BY batter
ORDER BY total_runs DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    batter,
    phase,
    SUM(batter_runs) AS runs,
    SUM(ball_faced) AS balls,
    ROUND(
        100.0 * SUM(batter_runs) / NULLIF(SUM(ball_faced), 0),
        2
    ) AS strike_rate,
    SUM(is_boundary) AS boundaries
FROM deliveries
GROUP BY batter, phase
HAVING SUM(ball_faced) >= 100
ORDER BY strike_rate DESC;
"""

phase_sql = pd.read_sql_query(query, conn)

phase_sql.head(15)
```

```python
query = """
SELECT
    batter,
    SUM(batter_runs) AS death_runs,
    SUM(ball_faced) AS balls,
    ROUND(
        100.0 * SUM(batter_runs) / NULLIF(SUM(ball_faced), 0),
        2
    ) AS strike_rate,
    SUM(is_boundary) AS boundaries,
    ROUND(
        100.0 * SUM(is_boundary) / NULLIF(SUM(ball_faced), 0),
        2
    ) AS boundary_pct
FROM deliveries
WHERE phase = 'Death'
GROUP BY batter
HAVING SUM(ball_faced) >= 300
ORDER BY strike_rate DESC
LIMIT 15;
"""

death_specialists = pd.read_sql_query(query, conn)

death_specialists
```

```python
query = """
SELECT
    batter,
    bowler,
    SUM(batter_runs) AS runs,
    SUM(ball_faced) AS balls,
    ROUND(
        100.0 * SUM(batter_runs) / NULLIF(SUM(ball_faced), 0),
        2
    ) AS strike_rate,
    SUM(is_boundary) AS boundaries,
    SUM(is_wicket) AS wicket_events
FROM deliveries
GROUP BY batter, bowler
HAVING SUM(ball_faced) >= 30
ORDER BY runs DESC
LIMIT 20;
"""

matchups = pd.read_sql_query(query, conn)

matchups
```

```python
query = """
SELECT
    batting_team,
    phase,
    SUM(total_runs) AS runs,
    SUM(is_legal_delivery) AS legal_balls,
    ROUND(
        100.0 * SUM(total_runs) /
        NULLIF(SUM(is_legal_delivery), 0),
        2
    ) AS scoring_rate,
    SUM(is_boundary) AS boundaries,
    SUM(is_wicket) AS wicket_events
FROM deliveries
GROUP BY batting_team, phase
ORDER BY batting_team, phase;
"""

team_phase = pd.read_sql_query(query, conn)

team_phase.head(20)
```
